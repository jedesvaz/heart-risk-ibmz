```python
from llama_cpp import Llama

llm = Llama(
    model_path="/home/ibm-user/granite-4.0-h-350m-base-be.f16.gguf",
    n_ctx=2048,
    n_threads=3,
    verbose=False,
)

def ask_llm(prompt, max_tokens=120):
    out = llm(prompt, max_tokens=max_tokens, temperature=0,
              stop=["\n\nSTATS:", "\n\n"])
    return out["choices"][0]["text"].strip()
```
